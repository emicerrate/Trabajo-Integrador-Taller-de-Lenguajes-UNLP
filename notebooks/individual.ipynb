```python
import csv
from src.constantes import DATA_PATH, DATA_OUT_PATH
from pathlib import Path
from src.funcionesI import trad_niv_ed
from glob import glob
```

```python
data_comp_path = DATA_PATH /"usu_individual.csv"
header_writen = False
with open(data_comp_path, "w") as data_comp:
    for archive in DATA_PATH.glob("usu_individual_*"):
        with open(archive) as data:
            if not header_writen:
                data_comp.write(next(data))
                header_writen = True
            else:
                next(data)
            for line in data:
                data_comp.write(line)
                
```

```python
data_ind_path = DATA_PATH /"usu_individual.csv"
dataout_ind_path = DATA_OUT_PATH /"usu_individual.csv" 
with open(data_ind_path, "r") as data_ind:
    data_ind = csv.reader(data_ind, delimiter=";")
    with open(dataout_ind_path, "w") as data_out:
        data_out_csv = csv.writer(data_out, delimiter=";")
        header = next(data_ind)
        header.append("NIVEL_ED_str")
        data_out_csv.writerow(header)
        column = header.index("NIVEL_ED")
        for line in data_ind:
            row = line[:] + [trad_niv_ed(line, column)]
            data_out_csv.writerow(row)
```
